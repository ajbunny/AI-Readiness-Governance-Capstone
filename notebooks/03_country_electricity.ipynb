## Creating a list of the selected countries to filter the electricity access data

```python
selected_countries = [
    "Ethiopia",
    "Uganda",
    "Rwanda",
    "Congo, Dem. Rep.",
    "Burkina Faso",
    "Madagascar",
    "Afghanistan",
    "Yemen, Rep.",
    "Kenya",
    "Angola",
    "India",
    "Cambodia",
    "Morocco",
    "Honduras",
    "Kyrgyz Republic",
    "Papua New Guinea",
    "Colombia",
    "Jamaica",
    "Armenia",
    "Ukraine",
    "Indonesia",
    "Samoa",
    "Iraq",
    "Botswana",
    "Japan",
    "Spain",
    "Barbados",
    "Israel",
    "United States"
]
```

```python
len(selected_countries)
```

```python
electricity_candidates = df[df["Country Name"].isin(selected_countries)]

electricity_candidates
```

```python
electricity_candidates.shape
```

```python
electricity_candidates.info()
```

```python
year_cols = [str(year) for year in range(2018, 2025)]

year_cols
```

```python
id_cols = ["Country Name", "Country Code", "Indicator Name", "Indicator Code"]
```

```python
id_cols + year_cols
```

```python
## Combine the id_cols and year_cols to create a new dataframe with only the relevant columns

electricity_candidates = electricity_candidates[id_cols + year_cols]
```
