```python
## Creating a list so incomes are from low -> high

income_order = ['Low income', 'Lower middle income', 'Upper middle income', 'High income']
```

```python
countries["Income group"] = pd.Categorical(
    countries['Income group'],
    categories=income_order,
    ordered=True
)
```

## Candidate Country Screening Pool

Countries were first grouped using the World Bank income classifications:
Low Income, Lower Middle Income, Upper Middle Income, and High Income.

From each income group, countries were selected from different regions to
create geographic diversity in the sample. More countries were selected from
the low- and middle-income groups because the main focus of this project is
understanding digital infrastructure and access in developing economies.
High-income countries were included as a smaller comparison group.

The initial screening pool includes 29 countries:

- 8 Low Income
- 8 Lower Middle Income
- 8 Upper Middle Income
- 5 High Income

These are not the final countries that will be used in the analysis. The next
step is to check how much ITU data is available for each country between
2018 and 2025. Countries with too much missing data may need to be replaced
or removed from the final sample.
